## Variable types in GreenLight

Our goal is to define a model which we will eventually run using the [GreenLight modelling platform](https://github.com/davkat1/greenlight). Variables in GreenLight have 4 possible **variable types**:
- `const` - variables that remain constant throughout a model run
- `input` - variable that are provided to the model as an input
- `aux` - an auxiliary variables: variables that are calculated based on other variables
- `state` - variables that are defined by ordinary differential equations (ODEs)
