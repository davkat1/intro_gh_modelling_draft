## Defining models in GreenLight

Models are defined in GreenLight using a [Python dictionary](https://docs.python.org/3/tutorial/datastructures.html#dictionaries) (`dict`) that defines the model variables. Each variable is itself defined using a dictionary, so the model definition is in fact a dictionary of dictionaries. The *keys* of the model definition dictionary are the names of the variables. For each variable, the corresponding *value* is another dictionary, with the following *keys*:
- `description` - a description of the variable
- `unit` - the unit that the variable's values are in
- `type` - the type of variable, which allows GreenLight to know how to process the variable: `state`, `const`, `input`, or `aux` (see [Lesson 1.2](../L1_a_simple_crop_model/102_variable_types.ipynb))
- `reference` - a reference describing where the variable definition came from
- `definition` - a definition of the variable. This may be a single value, input data, or a calculation based on other variables
- `init` - for variables of type state, an **initial value** which these variables have at the beginning of the model run

### The reference field in GreenLight

The goal of the reference filed in the GreenLight model definition is to provide readers with an explanation where the variable definition came from. This is typically a reference to literature, for example a book and page number or equation number. The reference may include any other reasoning about the choice of the variable definition. By including a reference with each variable definition, readers may check the origin of the definition, so they can see if it is implemented correctly, or reflect on whether or how they may change the definition.

### The definition field in GreenLight

In GreenLight model descriptions, variables should have a `definition` field which describes how the variable should be calculated. The `definition` field works as follows:
1. The `definition` is always defined with a string, even if it represents a number. So use for example `"definition": "6"` and not `"definition": 6`.
2. For `const` variables, the `definition` should be simply a number (example: `"definition": "6"`)
3. For `aux` variables, the `definition` is a mathematical expression involving other variables (example: `"definition": "1-exp(-k*LAI)"`)
4. For `input` variables, no definition is needed since the values should be provided as a data file. A constant value (like in `const`) can be provided as a default, in case no data is available.
5. For `state` variables, the `definition` is a mathematical expression describing the ordinary differential equation (ODE) defining the model (see example below)
